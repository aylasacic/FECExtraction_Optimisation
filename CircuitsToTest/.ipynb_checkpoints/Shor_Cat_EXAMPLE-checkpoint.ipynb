```python
from spidercat_shor import CatLibrary, MppCatRewriter
 
SPIDERCAT = "spidercat-main/spidercat"

EXAMPLE = """
MPP X0*X5*Y6*Y7*X8
MPP Z0*Z5*Z7*Z8*X9
MPP X1*Z4*Y5*Y6*X7*X8*Z9
MPP Z1*Z5*Z6*Z7*Y8*Y9
MPP X2*X5*Z6*X7*X8*Z9
MPP Z2*Z4*Y6*Y7*X8*X9
MPP X3*Z4*Y5*Z6*Z7*Z8*Y9
MPP Z3*Z5*X6*Y7*Y9
MPP X4*X5*X7*Y8*Y9
"""
```

```python
d = 2
T = d-1

lib = CatLibrary(SPIDERCAT, prefer="extract")
res = MppCatRewriter(lib, t=T, ancilla_mode="fresh").rewrite(EXAMPLE)

print(res.summary())
res.circuit.to_file("NAME.stim")

from StimToPyZX import mpp_to_cat_zx, save_zx

g, info = mpp_to_cat_zx(EXAMPLE, lib, t=T)
print(f"\n{g.num_vertices()} vertices, max arity "
      f"{max(g.vertex_degree(v) for v in g.vertices())}")
for h in info["cat_hubs"]:
    print(f"  weight {h['weight']:>2}  t={h['t']}  |V|={h['num_vertices']}  {h['source']}")

for p in save_zx(g, "NAME", png = True):
    print("saved:", p.resolve())

json_data = g.to_json()
with open("NAME.json", "w") as f:
    f.write(json_data)
# import pyzx as zx; zx.draw(g)
```
