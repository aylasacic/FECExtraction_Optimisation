## Big example

```python
k = CircuitExtractor(0)
k.split_zero_to_many(2,"Z")
k.split_zero_to_many(1,"Z")
k.split_zero_to_many(2,"Z")
# k.stretch_wire(2,0,0)
# k.split_one_to_many(0,2,"Z")
# k.merge_many_to_one([5,2],"Z")
k.add_measurement(1, 2, "Z")
k.add_measurement(2, 3, "Z")

k.split_zero_to_many(2,"Z")
k.split_zero_to_many(1,"Z")
k.split_zero_to_many(2,"Z")
k.add_measurement(8, 9, "Z")
k.add_measurement(9, 10, "Z")

k.split_zero_to_many(2,"Z")
k.split_zero_to_many(1,"Z")
k.split_zero_to_many(2,"Z")
k.add_measurement(15, 16, "Z")
k.add_measurement(16, 17, "Z")

k.split_zero_to_many(2,"Z")
k.split_zero_to_many(1,"Z")
k.split_zero_to_many(2,"Z")
k.add_measurement(22, 23, "Z")
k.add_measurement(23, 24, "Z")

k.split_zero_to_many(3,"Z")
k.split_zero_to_many(3,"Z")
k.split_zero_to_many(3,"Z")
k.split_zero_to_many(3,"Z")

k.split_zero_to_many(2,"Z")
k.split_zero_to_many(2,"Z")
k.add_measurement(41, 42, "Z")

k.split_zero_to_many(2,"Z")
k.split_zero_to_many(2,"Z")
k.add_measurement(46, 47, "Z")

k.stretch_wire(0,1,3)
k.merge_many_to_one([0,31],"X")
k.stretch_wire(1,2,3)
k.merge_many_to_one([1,34],"X")
k.stretch_wire(2,2,4)
k.merge_many_to_one([2,37],"X")
k.stretch_wire(3,3,4)
k.merge_many_to_one([3,40],"X")
k.stretch_wire(4,1,7)
k.merge_many_to_one([4,45],"X")

k.draw()
```
